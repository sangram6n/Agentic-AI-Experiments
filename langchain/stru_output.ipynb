```python
from langchain.tools import tool
from langchain.messages import HumanMessage, AIMessage
```

```python
@tool
def get_weather(city:str)->str:
    """ This the Weather finding function of a given city 
    given city it help to find weathe of that perticular city"""
    city = city.lower().strip()
    weather ={
        "puri": "its humid today",
        "bengaluru":"its slightly cold today",
        "bhubaneswar": "it is hot and humid today"
    }
    return weather.get(city,"I am Soory he weather of the city is not found Bimboo")

@tool
def additon(a:int,b:int) -> int:
    """this tool return addition of two numbers"""
    return a+b
```

```python
mode_t = groq_model.bind_tools([get_weather,additon])
```

```python
messages = []
query = HumanMessage(content="result  of 3+9 ")
messages.append(query)
ai_msg = mode_t.invoke(messages)
messages.append(ai_msg)
```

```python
for tool_call in ai_msg.tool_calls:
    print(tool_call["name"])
    if tool_call["name"] == "get_weather":
        tool_res = get_weather.invoke(tool_call)
    else:
        tool_res = additon.invoke(tool_call)
    messages.append(tool_res)
fin_ans = mode_t.invoke(messages)
print(fin_ans.text)
```

```python

from typing import List, Literal
from pydantic import BaseModel, Field
class Movie(BaseModel):
    Title : str = Field(...,description="the title of the asked movie")
    year: int = Field(...,description="year of the movie relesed")
    director:str = Field(...,description="director of the movie")
    ratings: float = Field(...,description="rating of the movie out of 10")
    type_of_movie : Literal["good","average","flop"]
    actors : List[Actor]
    geners : List[str]
    budget: float|None = Field(None, description = "Budget of the movie")
```

```python
mode_s = groq_model.with_structured_output(Movie,include_raw= True)
```

```python
res = mode_s.invoke("give me details about man of steel")
```

```python
res["parsed"].dict()
```

```python
from typing import TypedDict
```

```python
class Actor(TypedDict):
    name : str = Field(description="name of the actorr")
    role : str = Field(description="role of the actor")
    colour :str = Field(description="colour of the actor")
    age : int = Field(description="age of the actor")
                                                
class Movie(TypedDict):
    Title : str = Field(...,description="the title of the asked movie")
    year: int = Field(...,description="year of the movie relesed")
    director:str = Field(...,description="director of the movie")
    ratings: float = Field(...,description="rating of the movie out of 10")
    type_of_movie : Literal["good","average","flop"]
    actors : List[Actor]
    geners : List[str]
    budget: float|None = Field(None, description = "Budget of the movie")
```

```python
mode_s = groq_model.with_structured_output(Movie,include_raw= True)
res = mode_s.invoke("give me details about man of steel")

```

```python
class ContactInfo(TypedDict):
    """Contact information for a person."""
    name: str # The name of the person
    email: str # The email address of the person
    phone: str # The phone number of the person
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model = groq_model,
    response_format = ContactInfo
)
```

```python
result = agent.invoke({
    "messages": [{"role": "user", "content": "Extract contact info from: John Doe, mali@gmail.com, 9853200122"}]
})
```
