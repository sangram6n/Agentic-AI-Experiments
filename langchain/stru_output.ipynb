```python
from langchain.tools import tool
from langchain.messages import HumanMessage, AIMessage
```

```python
@tool
def get_weather(city:str)->str:
    """ This the Weather finding function of a given city 
    given city it help to find weathe of that perticular city"""
    city = city.lower().strip()
    weather ={
        "puri": "its humid today",
        "bengaluru":"its slightly cold today",
        "bhubaneswar": "it is hot and humid today"
    }
    return weather.get(city,"I am Soory he weather of the city is not found Bimboo")

@tool
def additon(a:int,b:int) -> int:
    """this tool return addition of two numbers"""
    return a+b
```

```python
mode_t = groq_model.bind_tools([get_weather,additon])
```

```python
messages = []
query = HumanMessage(content="result  of 3+9 ")
messages.append(query)
ai_msg = mode_t.invoke(messages)
messages.append(ai_msg)
```

```python
for tool_call in ai_msg.tool_calls:
    print(tool_call["name"])
    if tool_call["name"] == "get_weather":
        tool_res = get_weather.invoke(tool_call)
    else:
        tool_res = additon.invoke(tool_call)
    messages.append(tool_res)
fin_ans = mode_t.invoke(messages)
print(fin_ans.text)
```
